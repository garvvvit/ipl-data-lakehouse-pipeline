```sql
SELECT 
    batter,
    SUM(total_runs) AS career_runs,
    ROUND(SUM(total_runs) / NULLIF(SUM(dismissals), 0), 2) AS career_average,
    ROUND((SUM(total_runs) / NULLIF(SUM(balls_faced), 0)) * 100, 2) AS career_strike_rate,
    SUM(fours) AS total_fours,
    SUM(sixes) AS total_sixes
FROM workspace.default.gold_batter_stats
GROUP BY batter
ORDER BY career_runs DESC
LIMIT 10;
```

```sql
SELECT 
    venue,
    SUM(matches_played) AS total_matches,
    SUM(total_runs) AS total_venue_runs,
    ROUND(SUM(total_runs) / NULLIF(SUM(matches_played), 0), 2) AS avg_runs_per_match,
    ROUND(AVG(average_runs_per_ball) * 6, 2) AS avg_runs_per_over
FROM workspace.default.gold_venue_summary
GROUP BY venue
HAVING total_matches >= 5
ORDER BY avg_runs_per_match DESC
LIMIT 10;
```

```sql
WITH RankedVenues AS (
    SELECT 
        season,
        venue,
        matches_played,
        average_runs_per_match,
        ROW_NUMBER() OVER (PARTITION BY season ORDER BY average_runs_per_match DESC) AS venue_rank
    FROM workspace.default.gold_venue_summary
    WHERE matches_played >= 2
)
SELECT season, venue_rank, venue, average_runs_per_match
FROM RankedVenues
WHERE venue_rank <= 3
ORDER BY season DESC, venue_rank ASC;
```
